## Q3. Data Visualization

Create the following visualizations:

1.Class Distribution Plot

2.Correlation Heatmap

3.Histogram of selected features

4.Pair Plot (for any four important features)
Write your observations.

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# class Distribution plot
sns.countplot(x=y_series)
plt.title("Class Distribution")
plt.xlabel("Target Class")
plt.ylabel("Count")
plt.show()
```

Observation:Class 0 represents Malignant.Class 1 represents Benign.Benign cases (357) are higher than malignant cases (212).The dataset is moderately imbalanced.

```python
# Correlation Heatmap
import seaborn as sns
import matplotlib.pyplot as plt
data = df_x.copy()
data["target"] = y
plt.figure(figsize=(20,15))
sns.heatmap(data.corr(),annot=True ,cmap="coolwarm")
plt.title("Correlation Heatmap")
plt.show()
```

```python
# 3. Histogram of Selected Features

features = [
    "worst concave points",
    "mean concave points",
    "worst perimeter",
    "worst radius"
]

for feature in features:
    
    plt.figure(figsize=(7, 5))
    
    sns.histplot(
        data=data,
        x=feature,
        bins=20
        
    )
    
    plt.xlabel(feature)
    plt.ylabel("Count")
    plt.title(f"{feature} Distribution by Target")
    plt.show()
```

Observation:Mean radius and mean area show right-skewed distributions.Mean texture has a varied distribution.
Mean smoothness values are concentrated in a relatively narrow range.Histograms help understand the distribution of features.

```python
# Pair Plot (Four Important Features)
features = ["mean radius", "mean texture", "mean area", "mean smoothness"]

pair_data = df_x[features].copy()

pair_data["target"] = y_series

sns.pairplot(
    pair_data,
    vars=features,
    hue="target"
)

plt.suptitle("Pair Plot of Four Important Features", y=1.02)

plt.show()
```

Observation:-The pair plot shows relationships between the selected features. Mean radius, mean perimeter, and mean area have strong relationships with each other, while some feature combinations show better separation between the classes.

## Q4. Data Preprocessing

Prepare the dataset by:

1.Separating features (X) and target (y)

2.Splitting the data into 80% training and 20% testing sets

3.Applying feature scaling using StandardScaler (SKIP THIS STEP)

```python
# Separating features(X)and target(y)
X = df_x
y = y_series

# Split into training and testing data
from sklearn.model_selection import train_test_split

xtrain, xtest,ytrain, ytest = train_test_split(X,y,train_size = 0.8,random_state=21)
print(xtrain.shape,xtest.shape)
print(ytrain.shape,ytest.shape)
```

```python
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer

num_pipe = make_pipeline(SimpleImputer(strategy = 'median')).set_output(transform = "pandas")
num_pipe
```

```python
# fit it on training data
num_pipe.fit(xtrain)
```

```python
# Transform it on tesing as well as training data
xtrain_pre = num_pipe.transform(xtrain)
xtest_pre = num_pipe.transform(xtest)
xtrain_pre.head()
```

## Q5. Model Training
Build a Logistic Regression model using the training dataset and display the model coefficients and intercept.

```python
from sklearn.linear_model import LogisticRegression

# crate model
model = LogisticRegression()

# train model
model.fit(xtrain_pre,ytrain)
```

```python
# Coefficients
model.coef_
```

```python
# Intercept
model.intercept_
```

## Q6. Prediction

Predict the target values for the testing dataset and display a comparison of the Actual and Predicted values.

```python
ypred = model.predict(xtest_pre)

df_comparison = pd.DataFrame({
    'Actual': ytest.values,
    'Predicted': ypred
})

df_comparison
```

Q7. Model Evaluation

Evaluate the model using:

1.Accuracy Score

2.Precision Score

3.Recall Score

4.F1-Score

5.Confusion Matrix

6.Classification Report
Interpret the results.

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score

# Prediction
ytest_preds = model.predict(xtest_pre)
```

```python
# Accuracy
accuracy = accuracy_score(ytest, ytest_preds)
accuracy
```

```python
# Precision
precision = precision_score(ytest, ytest_preds)
precision
```

```python
# Recall
recall = recall_score(ytest, ytest_preds)
recall
```

```python
# F1 Score
from sklearn.metrics import f1_score

f1 = f1_score(ytest, ytest_preds)
f1
```

```python
# Confusion matrix
from sklearn.metrics import ConfusionMatrixDisplay
ConfusionMatrixDisplay.from_estimator(model,xtest_pre,ytest)
```

```python
# Classification report
from sklearn.metrics import classification_report

print(classification_report(ytest, ytest_preds))
```


### Interpretation of Results

* **Accuracy: 92%** — The model correctly classified 92% of the test samples.
* **Class 0:** Precision is 92%, Recall is 85%, and F1-score is 88%.
* **Class 1:** Precision is 92%, Recall is 96%, and F1-score is 94%.
* **Macro F1-score: 91%** — The overall performance across both classes is good.
* **Overall:** The model performed well, with a higher recall for Class 1 (96%), meaning it correctly identified most of the actual Class 1 cases.
