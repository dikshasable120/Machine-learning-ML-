## Assignment 2

```python
import pandas as pd

from sklearn.datasets import load_diabetes
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)
```

## Q1. Dataset Loading

Load the Diabetes Dataset using Scikit-learn and display the first 10 records, dataset shape, feature names, and target variable.

```python
df_X, series_y = load_diabetes(
    as_frame=True,
    return_X_y=True
)

X = df_X
y = series_y

print("First 10 Records:")
print(X.head(10))

print("\nDataset Shape:")
print(X.shape)

print("\nFeature Names:")
print(X.columns.tolist())

print("\nTarget Variable:")
print(y.head(10))
```

## Q2. Exploratory Data Analysis (EDA)

Perform Exploratory Data Analysis (EDA) by:

Checking data types

Identifying missing values

Displaying statistical summary

Finding duplicate records

```python
## Checking data types
print("nData Types:")
print(X.dtypes)
```

```python
# indentifying missing values
print("\nMissing Values:")
print(X.isnull().sum())
```

```python
# Displaying statistical summary
print("\nStatistical Summary:")
print(X.describe())
```

```python
# Finding duplicate records
print("\nDuplcated Records:")
print(X.duplicated().sum())
```

## Q3. Data Visualization

Create the following visualizations:

1.Histogram

2.Correlation Heatmap

3.Scatter Plot (between the most correlated feature and the target variable)

4.Write your observations for each visualization.

## Histogram

```python
import matplotlib.pyplot as plt
X.hist(figsize=(12, 10), bins=20)
plt.suptitle("Histogram of Diabetes Dataset Features")
plt.tight_layout()
plt.show()
```

## Correlation Heatmap

```python
import seaborn as sns
df = pd.concat([X, y.rename("target")], axis=1)

plt.figure(figsize=(12, 8))

sns.heatmap(
    df.corr(),
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Correlation Heatmap")
plt.show()
```

## Find Most Correlated Feature

```python
correlation = df.corr()["target"].drop("target")

most_correlated_feature = correlation.abs().idxmax()

print("\nMost Correlated Feature:")
print(most_correlated_feature)

print("Correlation Value:")
print(correlation[most_correlated_feature])
```

## Scatter Plot

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    x=df[most_correlated_feature],
    y=df["target"]
)

plt.xlabel(most_correlated_feature)
plt.ylabel("Disease Progression")
plt.title(
    f"{most_correlated_feature} vs Disease Progression"
)

plt.show()
```

## Observation

The scatter plot shows the relationship between the most correlated feature and disease progression. A positive correlation indicates that the target generally increases as the feature increases.

## Q4. Data Preprocessing

Prepare the dataset by:

1.Separating features (X) and target (y)

2.Splitting the data into 80% training and 20% testing sets

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)

print("\nTraining Data Shape:")
print(X_train.shape)

print("\nTesting Data Shape:")
print(X_test.shape)
```

## Q5. Model Training

Build a Linear Regression model using the training dataset and display the model coefficients and intercept.



```python
model = LinearRegression()

model.fit(X_train, y_train)

print("\nModel Coefficients:")

for feature, coefficient in zip(X.columns, model.coef_):
    print(feature, ":", coefficient)

print("\nModel Intercept:")
print(model.intercept_)
```

## Q6. Prediction

Predict the target values for the testing dataset and display a comparison of the Actual and Predicted values.

```python
y_pred = model.predict(X_test)

comparison = pd.DataFrame({
    "Actual": y_test.values,
    "Predicted": y_pred
})

print("\nActual vs Predicted:")
print(comparison)
```

## Q7. Model Evaluation

Evaluate the model using:

1.Mean Absolute Error (MAE)

2.Mean Squared Error (MSE)

3.Root Mean Squared Error (RMSE)

4.R² Score

Interpret the results.

```python
import numpy as np
mae = mean_absolute_error(y_test, y_pred)

mse = mean_squared_error(y_test, y_pred)

rmse = np.sqrt(mse)

r2 = r2_score(y_test, y_pred)

print("\nModel Evaluation")

print("MAE :", mae)
print("MSE :", mse)
print("RMSE:", rmse)
print("R2 Score:", r2)
```

## Interpretation

MAE, MSE, and RMSE measure the prediction errors of the model. Lower values indicate smaller errors. R² indicates how much variation in disease progression is explained by the model. A higher R² indicates better explanatory performance.

## Q9. Feature Analysis

Identify the most influential feature based on the regression coefficients and explain how it affects the target variable.

```python
coefficients = pd.DataFrame({
    "Feature": X.columns,
    "Coefficient": model.coef_
})

coefficients["Absolute_Coefficient"] = (
    coefficients["Coefficient"].abs()
)

coefficients = coefficients.sort_values(
    by="Absolute_Coefficient",
    ascending=False
)

print("\nFeature Analysis:")
print(coefficients)

most_influential = coefficients.iloc[0]

print("\nMost Influential Feature:")
print(most_influential["Feature"])

print("Coefficient:")
print(most_influential["Coefficient"])
```

## Explanation

## If the coefficient is positive:

The feature has a positive effect on the predicted target. As the feature increases, the predicted disease progression tends to increase, while keeping other features constant.

## If negative:

The feature has a negative relationship with the predicted target. As the feature increases, the predicted disease progression tends to decrease, while keeping other features constant.
