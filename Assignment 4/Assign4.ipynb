## Q1. Dataset Loading

Load the Wine Dataset using Scikit-learn and display the first 10 records, dataset shape, feature names, and target classes.

```python
import pandas as pd
from sklearn.datasets import load_wine


df_x,y_series =  load_wine(
    as_frame=True ,return_X_y=True
)
Y=pd.DataFrame(y_series)

# Create DataFrame with target column
df = df_x.copy()
df["target"] = y_series
```

```python
# display shape
df_x.shape
```

```python
# Feature names
df_x.columns
```

```python
# Target classes
df["target"].value_counts()
```
